```python
# 1、将用户输入的prompt，调用tokenizer.apply_chat_template() 进行tokenize，
# 2、将tokenize之后的input_ids，给到model.generate()，model.generate()是transformers库封装的自回归生成方法
# 3、将得到结果通过tokenizer进行一些解码，就得到了生成的文本
```

```python
from transformers import AutoTokenizer,AutoModelForCausalLM
model_path = "model/Qwen3-0.6B"
tokenizer = AutoTokenizer.from_pretrained(model_path)
model = AutoModelForCausalLM.from_pretrained(model_path)
```

```python
prompt = "你好，你是谁？"
message_list = [{"role":"user","content":prompt}]
res=tokenizer.apply_chat_template(message_list,add_generation_prompt = True,return_tensors = "pt")
print(res)
```

```python
input_ids = res["input_ids"]
attention_mask = res["attention_mask"]
generated_text = model.generate(inputs=input_ids,attention_mask = attention_mask,max_new_tokens = 500)
```

```python
decoded_text = tokenizer.decode(generated_text)
```
